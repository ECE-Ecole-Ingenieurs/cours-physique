# TP 01 — Cinématique du point matériel : Système de coordonnées

Dans ce TP, nous allons utiliser Python pour représenter un point matériel dans l'espace.

L'objectif est de comprendre une idée fondamentale en mécanique : **un même point matériel peut être décrit avec plusieurs systèmes de coordonnées.**

Nous allons partir des coordonnées cylindriques $(r, \theta, z)$, puis calculer les coordonnées cartésiennes correspondantes $(x, y, z)$. Enfin, nous visualiserons le même point dans l'espace.

> 💡 **Idée clé**
>
> Les coordonnées ne sont pas le point.
> Les coordonnées sont une manière de décrire la position du point.
>
> Le point matériel $M$ est le même, mais sa description peut changer selon le système de coordonnées choisi.

## 1. Rappel de cours

En mécanique du point matériel, on modélise un objet par un point $M$, repéré par le vecteur position $\overrightarrow{OM}$, où $O$ est l'origine du repère.

**Coordonnées cartésiennes** : le point est décrit par $M = (x, y, z)$, dans la base $(\overrightarrow{e_x}, \overrightarrow{e_y}, \overrightarrow{e_z})$ :

$$
\overrightarrow{OM} = x\,\overrightarrow{e_x} + y\,\overrightarrow{e_y} + z\,\overrightarrow{e_z}
$$

**Coordonnées cylindriques** : le même point est décrit par $M = (r, \theta, z)$, dans la base $(\overrightarrow{e_r}, \overrightarrow{e_\theta}, \overrightarrow{e_z})$ :

$$
\overrightarrow{OM} = r\,\overrightarrow{e_r} + z\,\overrightarrow{e_z}
$$

où :

- $r$ est la distance du point à l'axe vertical $z$ ;
- $\theta$ est l'angle dans le plan horizontal $(x, y)$ ;
- $z$ est la hauteur du point.

On passe de l'un à l'autre avec :

$$
x = r\cos(\theta), \qquad y = r\sin(\theta), \qquad z_{\mathrm{cart}} = z
$$

> ⚠️ **Attention — Base fixe ou base mobile ?**
>
> La base cartésienne $(\overrightarrow{e_x}, \overrightarrow{e_y}, \overrightarrow{e_z})$ est **fixe** : elle ne change pas quand $M$ se déplace.
>
> La base cylindrique $(\overrightarrow{e_r}, \overrightarrow{e_\theta}, \overrightarrow{e_z})$ est **mobile** : $\overrightarrow{e_r}$ et $\overrightarrow{e_\theta}$ suivent la position de $M$. C'est ce point, vu en détail au chapitre 1 du cours, qui complique le calcul de la vitesse et de l'accélération en coordonnées cylindriques.

## 3. Projection d'un vecteur dans la base cartésienne

Dans la base cartésienne $(\overrightarrow{e_x}, \overrightarrow{e_y}, \overrightarrow{e_z})$, les coordonnées d'un point sont les **composantes** de son vecteur position :

$$
\overrightarrow{OM} = x\,\overrightarrow{e_x} + y\,\overrightarrow{e_y} + z\,\overrightarrow{e_z}
$$

Les coordonnées $x$, $y$ et $z$ peuvent être interprétées comme les projections du vecteur $\overrightarrow{OM}$ sur les trois axes cartésiens.

Autrement dit :

- $x$ est la projection de $\overrightarrow{OM}$ sur l’axe $Ox$,

- $y$ est la projection de $\overrightarrow{OM}$ sur l’axe $Oy$,

- $z$ est la projection de $\overrightarrow{OM}$ sur l’axe $Oz$.

````python
def afficher_projections(x, y, z):
    """
    Plot the position vector OM and its projections onto the
    three Cartesian axes.
    """
    fig = plt.figure(figsize=(7, 6))
    ax = fig.add_subplot(111, projection="3d")

    ax.quiver(0, 0, 0, x, y, z, color="tab:red", arrow_length_ratio=0.08)
    ax.scatter(x, y, z, color="black", s=40)
    ax.text(x, y, z, "  M", fontsize=11)

    # Successive projections on the coordinate axes
    ax.plot([0, x], [0, 0], [0, 0], "--", color="tab:blue", label="Projection sur x")
    ax.plot([x, x], [0, y], [0, 0], "--", color="tab:orange", label="Projection sur y")
    ax.plot([x, x], [y, y], [0, z], "--", color="tab:green", label="Projection sur z")

    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.set_zlabel("z")
    ax.set_xlim(-5, 5)
    ax.set_ylim(-5, 5)
    ax.set_zlim(-5, 5)
    ax.set_box_aspect((1, 1, 1))
    ax.view_init(elev=22, azim=-58)
    ax.legend(loc="upper left")
    ax.set_title("Projections du vecteur position")
    fig.suptitle(
        rf"$\vec{{OM}} = {x:.1f}\,\vec{{e_x}} + {y:.1f}\,\vec{{e_y}} + {z:.1f}\,\vec{{e_z}}$",
        y=0.98,
    )

    plt.show()
    plt.close(fig)
````

````python
from ipywidgets import interact, FloatSlider

interact(
    afficher_projections,
    x=FloatSlider(value=3.0, min=-5.0, max=5.0, step=0.5),
    y=FloatSlider(value=2.0, min=-5.0, max=5.0, step=0.5),
    z=FloatSlider(value=1.5, min=-5.0, max=5.0, step=0.5),
);
````

> 🔎 **Interprétation**
>
> Le point matériel est repéré par un unique vecteur $\overrightarrow{OM}$.
> Les coordonnées cartésiennes $(x, y, z)$ indiquent simplement combien ce vecteur avance dans chaque direction de la base fixe.

## 4. Changement de base : cylindrique → cartésiennes

Les vecteurs de la base cylindrique $(\overrightarrow{e_r}, \overrightarrow{e_\theta})$ ne sont pas fixes comme $(\overrightarrow{e_x}, \overrightarrow{e_y})$ : ils tournent avec l'angle $\theta$. On peut les exprimer dans la base cartésienne :

$$
\overrightarrow{e_r} = \cos\theta\,\overrightarrow{e_x} + \sin\theta\,\overrightarrow{e_y}
\qquad
\overrightarrow{e_\theta} = -\sin\theta\,\overrightarrow{e_x} + \cos\theta\,\overrightarrow{e_y}
$$

C'est exactement la démonstration vue au chapitre 1 (dérivées des vecteurs de la base polaire).

**Travail demandé** : complétez la fonction `base_cylindrique` qui renvoie les vecteurs $\overrightarrow{e_r}$ et $\overrightarrow{e_\theta}$ exprimés dans la base cartésienne, pour un angle $\theta$ donné.

````python
def base_cylindrique(theta):
    '''
    Express the cylindrical basis vectors e_r and e_theta
    in the Cartesian basis, for a given angle theta.

    Parameters
    ----------
    theta : float
        Angular coordinate [rad].

    Returns
    -------
    e_r, e_theta : np.ndarray
        Basis vectors expressed in the Cartesian basis (2D, in the xy-plane).
    '''

    # Radial basis vector
    e_r = np.array([np.cos(theta), np.sin(theta)])

    # Theta basis vector
    e_theta = np.array([-np.sin(theta), np.cos(theta)])

    return e_r, e_theta
````

````python
def afficher_base_cylindrique(theta_degrees):
    """
    Plot the fixed Cartesian basis and the rotated cylindrical basis
    for a given angle theta.
    """
    theta = np.radians(theta_degrees)
    e_r, e_theta = base_cylindrique(theta)

    fig, ax = plt.subplots(figsize=(6, 6))

    # Fixed Cartesian basis
    ax.quiver(0, 0, 1, 0, color='grey', angles='xy', scale_units='xy', scale=1, label=r"$\vec{e_x}$")
    ax.quiver(0, 0, 0, 1, color='grey', angles='xy', scale_units='xy', scale=1, label=r"$\vec{e_y}$")

    # Mobile cylindrical basis
    ax.quiver(0, 0, *e_r, color='tab:orange', angles='xy', scale_units='xy', scale=1, label=r"$\vec{e_r}$")
    ax.quiver(0, 0, *e_theta, color='tab:blue', angles='xy', scale_units='xy', scale=1, label=r"$\vec{e_\theta}$")

    # Angle theta from the Cartesian x-axis to the radial vector
    arc_radius = 0.45
    angle_values = np.linspace(0.0, theta, 100)
    ax.plot( arc_radius * np.cos(angle_values),arc_radius * np.sin(angle_values), color='tab:red',linewidth=2)
    label_angle = theta / 2 if theta > 0 else 0.12
    ax.text( 0.58 * np.cos(label_angle), 0.58 * np.sin(label_angle), r"$\theta$", color='tab:red', fontsize=13)

    ax.set_xlim(-1.5, 1.5)
    ax.set_ylim(-1.5, 1.5)
    ax.set_aspect('equal')
    ax.axhline(0, color='black', linewidth=0.5)
    ax.axvline(0, color='black', linewidth=0.5)
    ax.set_title(rf"Base cylindrique pour $\theta = {theta_degrees:.0f}°$")
    ax.legend()
    plt.show()
    plt.close(fig)
````

````python
from ipywidgets import interact, FloatSlider

interact(
    afficher_base_cylindrique,
    theta_degrees=FloatSlider(value=35, min=0, max=360, step=5,  description=r'Theta (°)'),
);
````

> 💡 **Idée clé**
>
> Cette rotation est exactement ce qui complique le calcul de la vitesse et de
> l'accélération en coordonnées cylindriques : il faut aussi dériver
> $\overrightarrow{e_r}$ et $\overrightarrow{e_\theta}$ par rapport au temps,
> puisqu'ils tournent avec $\theta(t)$.

## 5. Changement de base : cartésiennes → cylindrique

On peut aussi faire la transformation inverse : exprimer les vecteurs fixes $\overrightarrow{e_x}$ et $\overrightarrow{e_y}$ en fonction des vecteurs mobiles $\overrightarrow{e_r}$ et $\overrightarrow{e_\theta}$, pour un angle $\theta$ donné :

$$
\overrightarrow{e_x} = \cos\theta\,\overrightarrow{e_r} - \sin\theta\,\overrightarrow{e_\theta}
\qquad
\overrightarrow{e_y} = \sin\theta\,\overrightarrow{e_r} + \cos\theta\,\overrightarrow{e_\theta}
$$

Comme il s'agit d'une rotation, l'inverse s'obtient simplement en changeant le signe de $\theta$ (ou, de manière équivalente, en transposant la matrice de rotation).

**Travail demandé** : complétez la fonction `base_cartesienne` qui renvoie les vecteurs $\overrightarrow{e_x}$ et $\overrightarrow{e_y}$ exprimés dans la base cylindrique, pour un angle $\theta$ donné.

````python
def base_cartesienne(theta):
    '''
    Express the Cartesian basis vectors e_x and e_y
    in the cylindrical basis, for a given angle theta.

    Parameters
    ----------
    theta : float
        Angular coordinate [rad].

    Returns
    -------
    e_x, e_y : np.ndarray
        Basis vectors expressed in the cylindrical basis
        (components along e_r, e_theta).
    '''

    # Cartesian x-axis expressed in the cylindrical basis
    e_x = np.array([np.cos(theta), -np.sin(theta)])

    # Cartesian y-axis expressed in the cylindrical basis
    e_y = np.array([np.sin(theta), np.cos(theta)])

    return e_x, e_y
````

````python
# Verification
# should reconstruct e_x = [1, 0] and e_y = [0, 1]
theta_test = np.radians(35)

e_r, e_theta = base_cylindrique(theta_test)
coeffs_x, coeffs_y = base_cartesienne(theta_test)

e_x_reconstruit = coeffs_x[0] * e_r + coeffs_x[1] * e_theta
e_y_reconstruit = coeffs_y[0] * e_r + coeffs_y[1] * e_theta

print("e_x reconstruit :", e_x_reconstruit, "(attendu : [1, 0])")
print("e_y reconstruit :", e_y_reconstruit, "(attendu : [0, 1])")
````

> 💡 **Idée clé**
>
> Cette vérification illustre une propriété générale des matrices de rotation :
> leur inverse est égale à leur transposée. Passer de la base cylindrique à la
> base cartésienne, puis revenir, redonne bien le point de départ.

## 6. Passage des coordonnées cylindriques aux coordonnées cartésiennes

On définit un point matériel $M$ par ses coordonnées cylindriques :

$$
r = 4~\mathrm{m}
$$

$$
\theta = 35^\circ
$$

$$
z = 3~\mathrm{m}
$$

Le vecteur position s'écrit $\overrightarrow{OM} = r\,\overrightarrow{e_r}$. En
utilisant l'expression de $\overrightarrow{e_r}$ vue à la section 4, on
retrouve directement :

$$
\overrightarrow{OM} = r\left( \cos\theta\,\overrightarrow{e_x} + \sin\theta\,\overrightarrow{e_y} \right)
= \underbrace{r\cos\theta}_{x}\,\overrightarrow{e_x} + \underbrace{r\sin\theta}_{y}\,\overrightarrow{e_y}
$$

d'où $x = r\cos\theta$ et $y = r\sin\theta$ : ce ne sont pas des formules à
apprendre par cœur, mais une conséquence directe du changement de base.

> ⚠️ **Attention — Unités**
>
> En Python, les fonctions `np.sin` et `np.cos` utilisent des angles en radians.
>
> Si l'angle est donné en degrés, il faut le convertir avec :
>
> ```python
> theta = np.radians(theta_degrees)
> ```

````python
# Cylindrical coordinates of the material point M
r = 4.0                    # radial distance [m]
theta_degrees = 35.0       # angle [deg]
z = 3.0                    # height [m]

# NumPy trigonometric functions use radians
theta = np.radians(theta_degrees)

print("Cylindrical coordinates")
print(f"r = {r:.3f} m")
print(f"theta = {theta_degrees:.3f} deg = {theta:.3f} rad")
print(f"z = {z:.3f} m")
````

**Travail demandé** : écrivez la fonction `cylindrical_to_cartesian` qui prend en entrée les coordonnées cylindriques $(r, \theta, z)$ d'un point matériel et qui renvoie ses coordonnées cartésiennes $(x, y, z)$.

````python
def cylindrical_to_cartesian(r, theta, z):
    '''
    Convert cylindrical coordinates to Cartesian coordinates.

    Parameters
    ----------
    r : float
        Radial distance from the z-axis [m].
    theta : float
        Angular coordinate in the xy-plane [rad].
    z : float
        Height coordinate [m].

    Returns
    -------
    x : float
        Cartesian coordinate along x [m].
    y : float
        Cartesian coordinate along y [m].
    z_cart : float
        Cartesian coordinate along z [m].
    '''

    # Projection of the radial distance along the x-axis
    x = r * np.cos(theta)

    # Projection of the radial distance along the y-axis
    y = r * np.sin(theta)

    # The vertical coordinate is unchanged
    z_cart = z

    return x, y, z_cart
````

## 7. Visualisation 3D

On applique maintenant la fonction de conversion pour obtenir les coordonnées cartésiennes de $M$, puis on visualise le même point dans l'espace.

La figure montre deux manières de lire la position du point :

- un repère cartésien, représenté par un cube de lecture ;
- une description cylindrique, représentée par un cylindre de rayon $r$ et de hauteur $z$.

Le point $M$ est unique : les deux descriptions correspondent au même point dans l'espace.

````python
def plot_point_cylindrical(r, theta, z):
    """
    Plot a point M described by cylindrical coordinates.
    """
    x, y, z_cart = cylindrical_to_cartesian(r, theta, z)

    fig, ax = plt.subplots(
        figsize=(9, 7),
        subplot_kw={"projection": "3d"}
    )

    # ------------------------------------------------------------------
    # Cylindrical guide
    # ------------------------------------------------------------------
    angle = np.linspace(0.0, 2.0 * np.pi, 100)

    x_bot, y_bot, z_bot = cylindrical_to_cartesian(r, angle,np.zeros_like(angle))
    x_top, y_top, z_top = cylindrical_to_cartesian(r, angle,z * np.ones_like(angle))

    # Bottom and top circles of the cylinder
    ax.plot(x_bot, y_bot, z_bot, linestyle=":", color='orange', linewidth=1)
    ax.plot(x_top, y_top, z_top, linestyle=":", color='orange', linewidth=1)

    # ------------------------------------------------------------------
    # Cartesian box
    # ------------------------------------------------------------------
    x_r = r * np.cos(theta)
    y_r = r * np.sin(theta)
    z_r = z
    ax.plot([0, x_r], [0, 0], [0, 0], linestyle=":", color='purple', linewidth=1.0)
    ax.plot([x_r, x_r], [0, y_r], [0, 0], linestyle=":", color='purple', linewidth=1.0)
    ax.plot([x_r, x_r], [y_r, y_r], [0, z_r], linestyle=":", color='purple', linewidth=1.0)


    # ------------------------------------------------------------------
    # Point and physical vectors
    # ------------------------------------------------------------------
    ax.scatter(x_r, y_r,  z_r, color='black',facecolors="none", s=160, linewidths=2, label=r"Point $M$ (Coord. cartésiennes)")
    ax.scatter(x, y, z_cart, color='orange',  s=80, label=r"Point $M$ (Coord. cylindriques)")

    # Position vector OM
    ax.plot([0, x_r], [0, y_r], [0, z_r], linewidth=2, color='black',  label=r"$\vec{OM}$ (Coord. cartésiennes)")
    ax.plot([0, x], [0, y], [0, z_cart],  linewidth=2, color='orange', label=r"$\vec{OM}$ (Coord. cylindriques)")

    # Cylindrical components
    ax.plot([0, x], [0, y], [0, 0], linestyle="--", linewidth=2, label=r"$r$")
    ax.plot([x, x], [y, y], [0, z_cart], linestyle="--", linewidth=2, label=r"$z$")

    # Angle theta in the xy-plane
    arc_radius = 0.3 * r
    angle_values = np.linspace(0.0, theta, 50)

    ax.plot(
        arc_radius * np.cos(angle_values),
        arc_radius * np.sin(angle_values),
        np.zeros_like(angle_values),
        linewidth=2,
        label=r"$\theta$"
    )

    # Text labels
    ax.text(0, 0, 0, "  O")
    ax.text(x, y, z_cart, "  M")

    # Axis limits
    LIMIT_XY = 9.0
    LIMIT_Z = 6.0
    ax.set_xlim(-LIMIT_XY, LIMIT_XY)
    ax.set_ylim(-LIMIT_XY, LIMIT_XY)
    ax.set_zlim(0, LIMIT_Z)
    ax.set_xticks(np.arange(-LIMIT_XY, LIMIT_XY + 1, 3))
    ax.set_yticks(np.arange(-LIMIT_XY, LIMIT_XY + 1, 3))
    ax.set_zticks(np.arange(0, LIMIT_Z + 1, 2))

    ax.set_xlabel(r"$x$ [m]")
    ax.set_ylabel(r"$y$ [m]")
    ax.set_zlabel(r"$z$ [m]")

    ax.set_title(
        rf"$M(r,\theta,z)=({r:.2f}, {theta:.2f}, {z:.2f})$"
        "\n"
        rf"$M(x,y,z)=({x:.2f}, {y:.2f}, {z_cart:.2f})$"
    )

    ax.legend(loc="center left", bbox_to_anchor=(1.05, 0.5))
    fig.tight_layout()
    plt.show()
    plt.close(fig)
````

````python
# Interactive version
def plot_point_cylindrical_degrees(r, theta_degrees, z):
    plot_point_cylindrical(r, np.radians(theta_degrees), z)

interact(
    plot_point_cylindrical_degrees,
    r=FloatSlider(value=4.0, min=0.5, max=8.0, step=0.1),
    theta_degrees=FloatSlider(value=35.0, min=0.0, max=360.0, step=5.0),
    z=FloatSlider(value=3.0, min=0, max=5.0, step=0.1),
);
````

> 🔎 **Interprétation**
>
> Le point physique n'a pas changé. Nous avons seulement changé sa description mathématique :
>
> $$
> (r, \theta, z) \longrightarrow (x, y, z)
> $$
