## 1. Rappel de cours

On étudie le mouvement d'un point matériel $M$ de masse $m$, lancé à
$t=0$ depuis une hauteur $h_0$. Sa vitesse initiale
$\overrightarrow{V_0}$ est caractérisée par sa norme $V_0$ et par
l'angle $\theta$ qu'elle forme avec l'horizontale.

Le mouvement est décrit dans le référentiel terrestre, supposé galiléen,
à l'aide du repère cartésien
$(O,\overrightarrow{u_x},\overrightarrow{u_y})$, où $(Ox)$ est
horizontal et $(Oy)$ vertical ascendant. Les relations cinématiques sont

$$
\overrightarrow{v}=\frac{\mathrm d\overrightarrow{OM}}{\mathrm dt}
\qquad\text{et}\qquad
\overrightarrow{a}=\frac{\mathrm d\overrightarrow{v}}{\mathrm dt}.
$$

Dans un référentiel galiléen, le principe fondamental de la dynamique
s'écrit

$$
\sum \overrightarrow{F}_{\mathrm{ext}}=m\overrightarrow{a}.
$$

Le point $M$ est soumis à son poids
$\overrightarrow{P}=-mg\,\overrightarrow{u_y}$. Selon le modèle
étudié, l'action de l'air est soit négligée, soit représentée par une
force de frottement opposée à la vitesse :

$$
\overrightarrow{f}=-\alpha\overrightarrow{v}
\qquad\text{ou}\qquad
\overrightarrow{f}=-\mu\|\overrightarrow{v}\|\overrightarrow{v},
$$

avec $\alpha>0$ et $\mu>0$. Pour chaque situation, il faudra établir
le bilan des forces, appliquer le PFD, le projeter sur les deux axes puis
déterminer le mouvement à partir des conditions initiales.

## 2. Balistique sans frottement

On considère d'abord un tir sans frottement de l'air. Le point $M$ est
lancé depuis $(0,h_0)$ avec une vitesse initiale
$\overrightarrow{V_0}$ de norme $V_0$, faisant un angle $\theta$ avec
l'horizontale. L'objectif est d'établir ses équations horaires, puis de
relier l'angle de tir à l'ensemble des points accessibles.

### 2.1 Théorie

:::{admonition} ✍️ Questions
:class: tip

1. Effectuer le bilan des forces, appliquer le PFD, puis déterminer les
   expressions des composantes $v_x(t)$ et $v_y(t)$ de la vitesse.
2. En déduire les équations horaires $x(t)$ et $y(t)$ en tenant compte
   des conditions initiales.
3. En déduire l'équation de la trajectoire $y(x)$.
4. À $V_0$ et $h_0$ fixés, on fait varier l'angle de tir $\theta$.
   Pour une abscisse $x$ donnée, déterminer la valeur de $\theta$ qui
   maximise $y(x)$, puis en déduire l'équation $y_{env}(x)$ de
   l'enveloppe de toutes les trajectoires possibles : la **parabole de
   sûreté**.
:::

:::{admonition} 💡 Aide
:class: hint

- Projetez le PFD sur $(Ox)$ et $(Oy)$.
- Pour la trajectoire, exprimez $t$ en fonction de $x$ à partir de
  $x(t)$, puis substituez cette expression dans $y(t)$.
- Pour la parabole de sûreté, posez $u=\tan\theta$ (donc
  $\cos^2\theta=\frac{1}{1+u^2}$). À $x$ fixé, $y$ s'exprime alors
  comme une fonction du second degré en $u$, dont vous pouvez chercher
  le maximum.
:::

:::{admonition} Réponses
:class: note

1. Le PFD projeté sur $(Ox)$ et $(Oy)$ donne $\dot v_x=0$ et
   $\dot v_y=-g$.
2. En intégrant, avec $v_x(0)=V_0\cos\theta$ et
   $v_y(0)=V_0\sin\theta$ :

   $$
   v_x(t)=V_0\cos\theta, \qquad v_y(t)=-gt+V_0\sin\theta
   $$

   puis, avec $x(0)=0$ et $y(0)=h_0$ :

   $$
   x(t)=V_0\cos\theta\,t, \qquad y(t)=-\frac12gt^2+V_0\sin\theta\,t+h_0
   $$
3. De $x(t)$, on tire $t=\dfrac{x}{V_0\cos\theta}$. En substituant
   dans $y(t)$ :

   $$
   y(x) = -\frac{g}{2V_0^2\cos^2\theta}x^2 + x\tan\theta + h_0
   $$
4. En posant $u=\tan\theta$, donc $\cos^2\theta=\dfrac1{1+u^2}$, on
   réécrit $y(x)$ à $x$ fixé comme une fonction de $u$ :

   $$
   y(u) = -\frac{g x^2}{2V_0^2}u^2 + xu + \left(h_0-\frac{gx^2}{2V_0^2}\right)
   $$

   C'est une fonction du second degré en $u$, à coefficient dominant
   négatif : son maximum est atteint pour
   $u^\star=\dfrac{V_0^2}{gx}$. En substituant :

   $$
   y_{env}(x) = h_0 + \frac{V_0^2}{2g} - \frac{g}{2V_0^2}x^2
   $$
:::

### 2.2 Implémentation numérique

:::{admonition} 💻 Code
:class: tip

Complétez deux fonctions :

1. `trajectoire_balistique`, qui renvoie l'altitude $y(x)$ établie à
   la question 3 ;
2. `parabole_surete`, qui renvoie l'altitude $y_{env}(x)$ établie à la
   question 4.
:::

```python
def trajectoire_balistique(x, v0, theta, h0):
    """
    Calcule l'altitude y(x) de la trajectoire balistique sans
    frottement.

    Paramètres
    ----------
    x : float ou np.ndarray
        Abscisse(s) considérée(s) [m].
    v0 : float
        Norme de la vitesse initiale [m/s].
    theta : float
        Angle de tir par rapport à l'horizontale [rad].
    h0 : float
        Hauteur initiale [m].

    Renvoie
    -------
    y : float ou np.ndarray
        Altitude y(x) [m].
    """
    g = 9.81
    y = -g / (2 * v0**2 * np.cos(theta)**2) * x**2 + x * np.tan(theta) + h0

    return y
```

```python
def parabole_surete(x, v0, h0):
    """
    Calcule l'altitude y_env(x) de la parabole de sûreté, enveloppe
    des trajectoires balistiques possibles à V0 et h0 fixés.

    Paramètres
    ----------
    x : float ou np.ndarray
        Abscisse(s) considérée(s) [m].
    v0 : float
        Norme de la vitesse initiale [m/s].
    h0 : float
        Hauteur initiale [m].

    Renvoie
    -------
    y_env : float ou np.ndarray
        Altitude y_env(x) [m].
    """
    g = 9.81
    y_env = h0 + v0**2 / (2 * g) - g / (2 * v0**2) * x**2

    return y_env
```

### 2.3 Analyse

Le code suivant représente la trajectoire associée à l'angle $\theta$
choisi, une famille de tirs réalisés avec la même norme $V_0$ et la
parabole de sûreté correspondante. Exécutez-le, puis faites varier $V_0$,
$h_0$ et $\theta$ pour analyser leur influence.

```python
def tracer_balistique(v0=20.0, h0=0.0, theta_deg=45.0):
    """Trace la trajectoire balistique, une famille d'angles, et la parabole de sûreté."""
    g = 9.81
    theta = np.radians(theta_deg)

    x_max = v0 / g * np.sqrt(v0**2 + 2 * g * h0)
    y_max = h0 + v0**2 / (2 * g)
    x_vals = np.linspace(0.0, x_max, 300)

    y_vals = trajectoire_balistique(x_vals, v0, theta, h0)
    y_env = parabole_surete(x_vals, v0, h0)

    fig, ax = plt.subplots(figsize=(9, 6))

    # Famille de trajectoires pour différents angles
    for theta_deg_i in np.arange(10, 90, 10):
        y_i = trajectoire_balistique(x_vals, v0, np.radians(theta_deg_i), h0)
        mask = y_i >= 0
        ax.plot(x_vals[mask], y_i[mask], color='0.8', linewidth=1)

    mask = y_vals >= 0
    ax.plot(x_vals[mask], y_vals[mask], color='tab:blue', linewidth=2, label=rf'trajectoire ($\theta={theta_deg:.0f}°$)')
    mask_env = y_env >= 0
    ax.plot(x_vals[mask_env], y_env[mask_env], color='tab:red', linestyle='--', linewidth=2, label='parabole de sûreté')
    ax.axhline(0, color='black', linewidth=0.5)
    ax.set_xlabel(r'$x$ (m)')
    ax.set_ylabel(r'$y$ (m)')
    ax.set_ylim(0, 1.05 * y_max)
    ax.set_title('Trajectoire balistique sans frottement')
    ax.grid(alpha=0.3)
    ax.legend(loc='upper right')

    plt.show()
    plt.close(fig)

tracer_balistique()
```

```python
interact(
    tracer_balistique,
    v0=FloatSlider(value=20.0, min=5.0, max=40.0, step=1.0, description='V0 (m/s)'),
    h0=FloatSlider(value=0.0, min=0.0, max=20.0, step=1.0, description='h0 (m)'),
    theta_deg=FloatSlider(value=45.0, min=5.0, max=85.0, step=5.0, description='Angle (°)'),
);
```

## 3. Balistique avec frottement de l'air

On reprend le même lancement, avec les mêmes conditions initiales, mais
l'action de l'air n'est plus négligée. Deux modèles sont étudiés : le
frottement linéaire, proportionnel à la vitesse, et le frottement
quadratique, proportionnel au carré de sa norme. L'objectif est de mettre
en équations ces deux modèles, puis de comparer numériquement les
trajectoires obtenues.

### 3.1 Théorie

:::{admonition} ✍️ Questions
:class: tip

1. On considère d'abord un frottement linéaire
   $\overrightarrow{f}=-\alpha\overrightarrow{v}$. Établir les
   équations différentielles vérifiées par $v_x(t)$ et $v_y(t)$, puis en
   déduire les expressions de $v_x(t)$, $v_y(t)$, $x(t)$ et $y(t)$.
2. On considère ensuite un frottement quadratique
   $\overrightarrow{f}=-\mu\|\overrightarrow{v}\|\overrightarrow{v}$.
   Établir les deux équations différentielles couplées vérifiées par
   $v_x(t)$ et $v_y(t)$, sans chercher à les résoudre.
:::

:::{admonition} 💡 Aide
:class: hint

- Pour le frottement linéaire, posez $\tau=m/\alpha$ et résolvez
  séparément les équations obtenues sur $(Ox)$ et $(Oy)$. Utilisez ensuite
  les conditions initiales pour intégrer la vitesse.
- Pour le frottement quadratique, écrivez
  $\|\overrightarrow{v}\|=\sqrt{v_x^2+v_y^2}$ avant de projeter le
  PFD. Cette norme fait intervenir simultanément les deux composantes.
:::

*Écrivez vos réponses ici (Markdown, $\LaTeX$…) :*

1. --
2. --

:::{admonition} Réponses
:class: note

1. Avec $\tau=m/\alpha$, le PFD projeté donne

   $$
   \dot v_x+\frac{1}{\tau}v_x=0,\qquad
   \dot v_y+\frac{1}{\tau}v_y=-g.
   $$

   En tenant compte de la vitesse initiale :

   $$
   v_x(t)=V_0\cos\theta\,e^{-t/\tau},
   $$

   $$
   v_y(t)=\left(V_0\sin\theta+g\tau\right)e^{-t/\tau}-g\tau.
   $$

   Une nouvelle intégration, avec $x(0)=0$ et $y(0)=h_0$, conduit à

   $$
   x(t)=\tau V_0\cos\theta\left(1-e^{-t/\tau}\right),
   $$

   $$
   y(t)=h_0+\tau\left(V_0\sin\theta+g\tau\right)
   \left(1-e^{-t/\tau}\right)-g\tau t.
   $$
2. Pour le frottement quadratique, le PFD projeté sur les deux axes donne

   $$
   \dot v_x=-\frac{\mu}{m}\sqrt{v_x^2+v_y^2}\,v_x,
   $$

   $$
   \dot v_y=-g-\frac{\mu}{m}\sqrt{v_x^2+v_y^2}\,v_y.
   $$

   Les deux équations sont couplées par la norme de la vitesse.
:::

### 3.2 Implémentation numérique

:::{admonition} 💻 Code
:class: tip

Complétez les fonctions `trajectoire_euler_lineaire` et
`trajectoire_euler_quadratique`. Pour chacune d'elles :

- initialisez les tableaux $x$, $y$, $v_x$ et $v_y$ à partir des
  conditions initiales ;
- à chaque pas, calculez les composantes de l'accélération ;
- appliquez la méthode d'Euler explicite aux quatre variables en utilisant
  uniquement leurs valeurs au pas $i$ ;
- renvoyez les tableaux contenant $x(t)$ et $y(t)$.
:::

```python
def trajectoire_euler_lineaire(t_vals, v0, theta, h0, m, alpha):
    """Intègre la trajectoire avec frottement linéaire par Euler."""
    g = 9.81
    h = t_vals[1] - t_vals[0]
    x_vals = np.zeros_like(t_vals, dtype=float)
    y_vals = np.zeros_like(t_vals, dtype=float)
    vx_vals = np.zeros_like(t_vals, dtype=float)
    vy_vals = np.zeros_like(t_vals, dtype=float)

    y_vals[0] = h0
    vx_vals[0] = v0 * np.cos(theta)
    vy_vals[0] = v0 * np.sin(theta)

    for i in range(len(t_vals) - 1):
        ax = -(alpha / m) * vx_vals[i]
        ay = -g - (alpha / m) * vy_vals[i]

        x_vals[i + 1] = x_vals[i] + h * vx_vals[i]
        y_vals[i + 1] = y_vals[i] + h * vy_vals[i]
        vx_vals[i + 1] = vx_vals[i] + h * ax
        vy_vals[i + 1] = vy_vals[i] + h * ay

    return x_vals, y_vals
```

```python
def trajectoire_euler_quadratique(t_vals, v0, theta, h0, m, mu):
    """Intègre la trajectoire avec frottement quadratique par Euler."""
    g = 9.81
    h = t_vals[1] - t_vals[0]
    x_vals = np.zeros_like(t_vals, dtype=float)
    y_vals = np.zeros_like(t_vals, dtype=float)
    vx_vals = np.zeros_like(t_vals, dtype=float)
    vy_vals = np.zeros_like(t_vals, dtype=float)

    y_vals[0] = h0
    vx_vals[0] = v0 * np.cos(theta)
    vy_vals[0] = v0 * np.sin(theta)

    for i in range(len(t_vals) - 1):
        norme_v = np.hypot(vx_vals[i], vy_vals[i])
        ax = -(mu / m) * norme_v * vx_vals[i]
        ay = -g - (mu / m) * norme_v * vy_vals[i]

        x_vals[i + 1] = x_vals[i] + h * vx_vals[i]
        y_vals[i + 1] = y_vals[i] + h * vy_vals[i]
        vx_vals[i + 1] = vx_vals[i] + h * ax
        vy_vals[i + 1] = vy_vals[i] + h * ay

    return x_vals, y_vals
```

### 3.3 Analyse

Le code suivant compare, pour un même lancement, les trajectoires sans
frottement, avec frottement linéaire et avec frottement quadratique.
Exécutez-le, puis faites varier $\alpha$ et $\mu$ afin d'observer
l'influence de chaque modèle.

```python
def tracer_comparaison_frottements(alpha=0.3, mu=0.02):
    """Compare trois modèles de trajectoire balistique."""
    g = 9.81
    m = 1.0
    v0 = 20.0
    h0 = 0.0
    theta = np.radians(45.0)
    t_vals = np.linspace(0.0, 8.0, 1601)

    x_sans = v0 * np.cos(theta) * t_vals
    y_sans = h0 + v0 * np.sin(theta) * t_vals - 0.5 * g * t_vals**2
    x_lin, y_lin = trajectoire_euler_lineaire(
        t_vals, v0, theta, h0, m, alpha
    )
    x_quad, y_quad = trajectoire_euler_quadratique(
        t_vals, v0, theta, h0, m, mu
    )

    fig, ax = plt.subplots(figsize=(9, 6))
    for x, y, couleur, style, etiquette in (
        (x_sans, y_sans, '0.4', '--', 'sans frottement'),
        (x_lin, y_lin, 'tab:blue', '-', 'frottement linéaire'),
        (x_quad, y_quad, 'tab:red', '-', 'frottement quadratique'),
    ):
        masque = y >= 0.0
        ax.plot(x[masque], y[masque], color=couleur, linestyle=style,
                linewidth=2, label=etiquette)

    ax.axhline(0, color='black', linewidth=0.7)
    ax.set_xlim(left=0)
    ax.set_ylim(bottom=0)
    ax.set_xlabel(r'$x$ (m)')
    ax.set_ylabel(r'$y$ (m)')
    ax.set_title('Influence du modèle de frottement sur la trajectoire')
    ax.grid(alpha=0.3)
    ax.legend()
    plt.show()
    plt.close(fig)

tracer_comparaison_frottements()
```

## 4. Application : trajectoire d'un volant de badminton

On envisage un volant de badminton envoyé depuis le fond du court avec
une vitesse initiale $\overrightarrow{V_0}$ formant un angle $\theta$
avec l'horizontale. Le volant est assimilé à un point matériel afin
d'étudier l'influence de l'angle de tir et de l'action de l'air sur sa
trajectoire.

### 4.1 Théorie

:::{admonition} 💡 Modèles et paramètres
:class: hint

Les frottements de l'air peuvent être modélisés par :

- une force de frottement linéaire
  $\overrightarrow{f}=-6\pi r\eta\,\overrightarrow{v}$. Dans les
  notations précédentes, $\alpha=6\pi r\eta$, où $\eta$ est la
  viscosité dynamique de l'air et $r$ une longueur caractéristique du
  volant ;
- une force de frottement quadratique
  $\overrightarrow{f}=-\mu\|\overrightarrow{v}\|\overrightarrow{v}$,
  avec $\mu=\frac12\rho S C_x$, où $\rho$ est la masse volumique de
  l'air, $S$ la surface frontale et $C_x$ le coefficient aérodynamique.

Pour commencer l'étude, utilisez le jeu de paramètres suivant :

$$
m=5{,}0\times10^{-3}\;\mathrm{kg},\quad
h_0=1{,}5\;\mathrm{m},\quad
r=3{,}3\times10^{-2}\;\mathrm{m},
$$

$$
\eta=1{,}8\times10^{-5}\;\mathrm{Pa\,s},\quad
\rho=1{,}2\;\mathrm{kg\,m^{-3}},\quad
S=\pi r^2,\quad C_x=0{,}1.
$$
:::

  figure.
:::

```python
def tracer_badminton(v0=30.0, theta_deg=30.0, Cx=0.1):
    """Compare les modèles de trajectoire pour un volant de badminton."""
    g = 9.81
    m = 5.0e-3
    h0 = 1.5
    r = 3.3e-2
    eta = 1.8e-5
    rho = 1.2
    S = np.pi * r**2
    alpha = 6 * np.pi * r * eta
    mu = 0.5 * rho * S * Cx
    theta = np.radians(theta_deg)

    t_sans = (
        v0 * np.sin(theta)
        + np.sqrt((v0 * np.sin(theta))**2 + 2 * g * h0)
    ) / g
    t_vals = np.linspace(0.0, 2.0 * t_sans, 2001)

    x_sans = v0 * np.cos(theta) * t_vals
    y_sans = h0 + v0 * np.sin(theta) * t_vals - 0.5 * g * t_vals**2
    x_lin, y_lin = trajectoire_euler_lineaire(
        t_vals, v0, theta, h0, m, alpha
    )
    x_quad, y_quad = trajectoire_euler_quadratique(
        t_vals, v0, theta, h0, m, mu
    )

    x_env_max = v0 / g * np.sqrt(v0**2 + 2 * g * h0)
    x_env = np.linspace(0.0, x_env_max, 400)
    y_env = parabole_surete(x_env, v0, h0)

    fig, ax = plt.subplots(figsize=(9, 6))
    for x, y, couleur, style, etiquette in (
        (x_sans, y_sans, '0.4', '--', 'sans frottement'),
        (x_lin, y_lin, 'tab:blue', '-', 'frottement linéaire'),
        (x_quad, y_quad, 'tab:red', '-', 'frottement quadratique'),
    ):
        masque = y >= 0.0
        ax.plot(x[masque], y[masque], color=couleur, linestyle=style,
                linewidth=2, label=etiquette)

    ax.plot(x_env, y_env, color='tab:orange', linestyle=':', linewidth=2,
            label='parabole de sûreté')
    ax.axhline(0, color='black', linewidth=0.7)
    ax.set_xlim(0, x_env_max)
    ax.set_ylim(0, 1.05 * (h0 + v0**2 / (2 * g)))
    ax.set_xlabel(r'$x$ (m)')
    ax.set_ylabel(r'$y$ (m)')
    ax.set_title('Trajectoire d’un volant de badminton')
    ax.grid(alpha=0.3)
    ax.legend()
    plt.show()
    plt.close(fig)

tracer_badminton()
```
